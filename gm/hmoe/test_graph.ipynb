```python
import ipycytoscape
import ipywidgets as widgets
from IPython.display import display

# Настройки таблицы
rows = 3
cols = 4
x_gap = 150  # расстояние между колонками
y_gap = 100  # расстояние между строками
intermediate_gap = 50  # промежуточные узлы между колонками

nodes = []
edges = []
experts_dist = [[3, 2, 4], [1, 2, 3], [3, 4, 1], [2, 3, 4]]

# Создаем основной узлы по таблице
for r in range(rows):
    for c in range(cols):
        node_id = f"N{r}_{c}"
        nodes.append({
            'data': {'id': node_id, 'label': f'E{experts_dist[c][r]}'},
            'position': {'x': c*(x_gap+intermediate_gap), 'y': r*y_gap},
            'classes': 'table-node'
        })

# Создаем промежуточные узлы между колонками

for c in range(cols + 1):
    inter_id = f"I{c}"
    inter_x = c*x_gap + (c+0.5)*intermediate_gap + x_gap
    inter_y = r*y_gap
    nodes.append({
        'data': {'id': inter_id, 'label': f'latent_{c}'},
        'position': {'x': inter_x, 'y': inter_y},
        'classes': 'inter-node'
    })

    for r in range(rows):
        # Соединяем основной узел с промежуточным
        if c > 0:
            edges.append({
                'data': {'id': f"{r}_{c}_to_I", 'source': f"N{r}_{c-1}", 'target': inter_id}
            })
        
        if c < cols:
            # Соединяем промежуточный с узлом следующей колонки
            edges.append({
                'data': {'id': f"I_to_{r}_{c+1}", 'source': inter_id, 'target': f"N{r}_{c}"}
            })

# Создаем виджет
cy = ipycytoscape.CytoscapeWidget(
    layout={'name': 'preset'},
    style={'width': '800px', 'height': '400px'}
)
cy.graph.add_graph_from_json({'nodes': nodes, 'edges': edges})

# Добавляем базовые стили
cy.set_style([
    {
        'selector': '.table-node',
        'style': {
            'background-color': '#3498db',
            'label': 'data(label)',
            'width': 40,
            'height': 40,
            'color': 'white',
            'text-valign': 'center',
            'text-halign': 'center'
        }
    },
    {
        'selector': '.inter-node',
        'style': {
            'background-color': '#95a5a6',
            'label': 'data(label)',
            'width': 50,
            'height': 50,
            'color': 'red',
        }
    },
    {
        'selector': 'edge',
        'style': {
            'line-color': '#2c3e50',
            'width': 2,
            'target-arrow-shape': 'triangle',
            'target-arrow-color': '#2c3e50',
            'curve-style': 'bezier'
        }
    }
])

display(cy)
```
